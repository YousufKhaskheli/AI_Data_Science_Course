### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
def genre_group(g):
    has_drama = 'Drama' in g
    has_action = 'Action' in g
    if has_drama and has_action:
        return 'Both'
    elif has_drama:
        return 'Drama'
    elif has_action:
        return 'Action'
    return None

movies['Genre_Group'] = movies['Genre'].apply(genre_group)
subset = movies[movies['Genre_Group'].isin(['Drama', 'Action', 'Both'])]
subset.boxplot(column='Gross', by='Genre_Group')
plt.title('Gross Revenue: Drama vs Action')
plt.suptitle('')
plt.xlabel('Genre')
plt.ylabel('Gross Revenue')
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.
