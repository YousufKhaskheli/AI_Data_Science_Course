## Merging: Joins

### Explanation of `pd.merge(how=...)`

The `how` parameter in `pd.merge()` dictates which rows are kept from the left and right DataFrames based on the common key(s):

- **`how='inner'` (default)**: Produces only the set of records that match in both DataFrames. It returns a DataFrame containing all rows from `data0` and `data2` that have matching `sales_man_id` values. Rows that do not have a match in both DataFrames are discarded.

- **`how='outer'`**: Produces the set of all records in both DataFrames, matching rows where possible, and filling in `NaN` (Not a Number) for unmatched fields. This means it includes all rows from `data0` and all rows from `data2`, and where there's no match for `sales_man_id`, the columns from the non-matching DataFrame will have `NaN` values.

- **`how='left'`**: Produces all records from the left DataFrame (`data0`) and any matching records from the right DataFrame (`data2`). If there is no match in `data2` for a row in `data0`, the columns from `data2` will be filled with `NaN`.

- **`how='right'`**: Produces all records from the right DataFrame (`data2`) and any matching records from the left DataFrame (`data0`). If there is no match in `data0` for a row in `data2`, the columns from `data0` will be filled with `NaN`.

```python
pd.merge(data0,data2,how='inner',on='sales_man_id')
```

```python
pd.merge(data0,data2, how='outer', on='sales_man_id')
```

```python
pd.merge(data0,data2, how='left', on='sales_man_id')
```

```python
pd.merge(data0,data2, how='right', on='sales_man_id')
```
