```python
import numpy as np
import pandas as pd
from datetime import datetime
import yfinance
import pandas_datareader
import matplotlib.pyplot as plt
```

```python
apple = yfinance.download('AAPL', start='2025-07-10', end='2026-07-10')
google = yfinance.download('GOOGL', start='2025-07-10', end='2026-07-10')
```

```python
apple
```

```python
apple_close = apple['Close']
google_close = google['Close']
```

```python
plt.figure(figsize=(16,4))
plt.plot(apple_close, label='Apple')
plt.plot(google_close, label='Google')
plt.legend()
plt.grid()
plt.show()
```

```python
apple_weekly = apple_close.resample('W').mean()
google_weekly = google_close.resample('W').mean()
plt.plot(apple_weekly, label="Apple")
plt.plot(google_weekly, label="Google")
plt.legend()
plt.grid()
plt.show()
```

```python
stocks = ['AAPL','GOOGL','AMZN','MSFT','NVDA']
closing_price = pd.DataFrame()
for stock in stocks:
    closing_price[stock] = yfinance.download(stock, start='2025-07-10', end='2026-07-10')['Close']
closing_price
```

```python
plt.figure(figsize=(16,4))
plt.plot(closing_price)
plt.title("Comparative Plots of Different IT Company Stocks of Clashing Price")
plt.legend(stocks)
plt.grid()
plt.show()
```
