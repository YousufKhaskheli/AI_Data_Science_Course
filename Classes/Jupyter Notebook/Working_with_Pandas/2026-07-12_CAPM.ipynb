#Capital Asset Pricing Model (CAPM)

```python
import numpy as np
import pandas as pd
import datetime
import yfinance as yf
import pandas_datareader as web
import plotly.express as px
```

```python
year = 2
```

```python
end = datetime.datetime.today()
end
```

```python
start = datetime.date(datetime.date.today().year - year, datetime.date.today().month, datetime.date.today().day)
start
```

```python
sp500 = web.DataReader(['sp500'], 'fred', start, end)
sp500.head()
```

```python
stocks_list = ['TSLA','AAPL','AMZN','GOOGL']
```

```python
stocks_df = pd.DataFrame()
for stock in stocks_list:
  data = yf.download(stock, period=f'{year}y')
  stocks_df[stock] = data['Close']
stocks_df.head()
```

```python
stocks_df.reset_index(inplace=True)
stocks_df
```

```python
sp500.reset_index(inplace=True)
sp500
```

```python
#sp500.rename(columns={'DATE':'Date'}, inplace=True)
```

```python
#data = pd.merge(stocks_df, sp500,on='Date', how='inner')
data = pd.merge(stocks_df, sp500,left_on='Date', right_on='DATE', how='inner').drop('DATE',axis=1)
data.head()
```

```python
data.set_index('Date',inplace=True)
```

```python
data.index
```

```python
#Functions to plot interactive plotly charts
def interactive_plot(df):
  fig = px.line()
  for i in df.columns:
    fig.add_scatter(x=df.index, y=df[i], name=i)
  fig.update_layout(
      width=800,
      margin=dict(l=20, r=20, t=50, b=20),
      legend=dict(x=1, xanchor="right", y=1.02, yanchor="bottom",
                  orientation='h')
  )
  return fig
```

```python
interactive_plot(data)
```
