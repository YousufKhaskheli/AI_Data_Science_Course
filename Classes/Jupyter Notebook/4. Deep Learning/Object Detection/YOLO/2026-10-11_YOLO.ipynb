```python
from ultralytics import YOLO   #  you only look once
```

```python
model = YOLO("yolo11n.pt")
```

```python
pred = model.predict("/content/pizza.jpg")
```

```python
pred[0].plot()
```

```python
img = model.predict("/content/img.jpg")
```

```python
img[0].show()
```

```python
import gradio as gr
def pred_image(img):
  img = model.predict(img)
  img = img[0].plot()
  return img
```

```python
gr.Interface(fn = pred_image, inputs ="image" , outputs = "image").launch()
```
