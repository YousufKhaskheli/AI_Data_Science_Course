## Lower/Upper

```python
text = """Natural         Lang!"#$%&'()*+,-./:;<=>?@[\]^_`{|}~u45555age       Proes555😚sing
'a',
 'about',
 'above',
 'after',
 'again',
 'against',
 'ain',
 'all',
 'am', (NLP) helps comput55ers understand, interpret and produce human language. It   studies langua👀ge as data and developsa model that can analyse linguistic   structure, meaning and context in both wr💔itten and spoken comm😩unication."""
```

```python
text = text.lower()
```

```python
text
```

## Digits Removed

```python
for digits in '1234567890':
  text = text.replace(digits,'')
```

### RE (Regular Expression)


```python
import re
text = re.sub('\d',"",text)
```

```python
text
```

### Extra Space Removed

```python
text = re.sub("\s+"," ",text)
```

```python
text
```

## Emojis Removed

```python
!pip install emoji
```

```python
text = text.encode("ascii","ignore").decode("ascii")
```

```python
text
```

## Punctuations (/,@,*,&,^,%)

```python
import string

print(string.punctuation)
```

```python
for punctuation in string.punctuation:
  text = text.replace(punctuation,'')
```

```python
text
```

## Stopwords Removed

```python
import nltk # Natural Language Tool kit
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
stop_words = stopwords.words("english")
```

```python
text
```

```python
text = " ".join([a for a in text.split() if a not in stop_words])
```
