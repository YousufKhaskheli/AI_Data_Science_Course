# Polynomial Regression

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn
from sklearn.preprocessing import OneHotEncoder
```

```python
data = pd.read_csv("insurance.csv")
```

```python
data.head()
```

```python
data.info()
```

```python
data.isna().sum()
```

```python
data.bmi.fillna(data.bmi.mean(),inplace=True)
```

```python
data.duplicated().sum()
```

```python
data.drop_duplicates(inplace=True); data.reset_index(drop=True, inplace=True)
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder()
encoded_categories = ohe.fit_transform(data.select_dtypes('O'))
```

```python
encoded_categories = encoded_categories.toarray()
```

```python
encoded_categories = pd.DataFrame(data=encoded_categories, columns=ohe.get_feature_names_out())
```

```python
numbered_data = data.select_dtypes('number')
numbered_data
```

```python
df = pd.concat([numbered_data,encoded_categories],axis=1)
```

```python
df
```

```python
df.dropna(inplace=True, axis=0)
df
```

```python
X = df.drop('charges', axis=1)
y = df.charges
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2, include_bias=True)
X_poly = poly.fit_transform(X)
```

```python
X = pd.DataFrame(data=X_poly, columns=poly.get_feature_names_out())
X
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
from sklearn.metrics import r2_score
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
y_pred_train = lr.predict(X_train)
```

```python
r2_score(y_train, y_pred_train)
```
