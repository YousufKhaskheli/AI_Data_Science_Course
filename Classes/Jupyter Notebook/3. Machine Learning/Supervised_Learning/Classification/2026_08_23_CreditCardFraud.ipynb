### EDA: Categorical and Boolean Column Analysis

```python
categorical_cols = data.select_dtypes(include=['object', 'bool']).columns

for col in categorical_cols:
    print(f'\nUnique values and counts for: {col}')
    display(data[col].value_counts().to_frame())
```

### EDA: Numerical Feature Distributions

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Distribution of Transaction Amount (amount_usd)
plt.figure(figsize=(10, 6))
sns.histplot(data['amount_usd'], bins=50, kde=True)
plt.title('Distribution of Transaction Amount (amount_usd)')
plt.xlabel('Amount (USD)')
plt.ylabel('Frequency')
plt.xlim(0, data['amount_usd'].quantile(0.99)) # Limiting x-axis for better visibility due to outliers
plt.show()
```

```python
# Distribution of amount_usd by is_fraud
plt.figure(figsize=(12, 7))
sns.histplot(data=data, x='amount_usd', hue='is_fraud', bins=50, kde=True, log_scale=True, palette='viridis')
plt.title('Distribution of Transaction Amount (log scale) by Fraud Status')
plt.xlabel('Amount (USD) (Log Scale)')
plt.ylabel('Frequency')
plt.show()
```
